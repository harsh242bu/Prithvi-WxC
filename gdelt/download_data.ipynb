```python
import re
import requests
import pandas as pd
import zipfile
import io
from bs4 import BeautifulSoup
```

```python
gdelt_url = "http://data.gdeltproject.org/events/index.html"
```

```python
response = requests.get(gdelt_url)
```

```python
response
```

```python
response.text
```

```python
# Parse the HTML
soup = BeautifulSoup(response.text, "html.parser")

# Extract all <a> tags
links = soup.find_all("a")

# Define a regex pattern to extract file details
pattern = re.compile(r'(?P<filename>\d+\.export\.CSV\.zip)"?>.*?\((?P<size>\d+\.\d+MB|\d+\.\d+GB)\).*?MD5:\s(?P<md5>[a-fA-F0-9]{32})')

# List to store extracted data
data = []

for link in links:
    match = pattern.search(str(link.parent))
    if match:
        data.append(match.groupdict())

# Convert to DataFrame
df = pd.DataFrame(data)

# Display the extracted data
df.head()
```

```python
output_dir = "/home/harsh242/pc_data/academics/academics_bu_grs/CS598-Multi-modal-ML/Project/Prithvi-WxC/gdelt_data"
```

```python
from urllib.parse import urljoin

zip_links = soup.find_all('a', href=lambda href: href and href.endswith('.zip'))

# Extract and print the full URLs of the ZIP files
for link in zip_links:
    zip_url = urljoin(gdelt_url, link['href'])
    zip_response = requests.get(zip_url)

    with zipfile.ZipFile(io.BytesIO(zip_response.content)) as z:
        for file_name in z.namelist():
            # if file_name.endswith('.csv'):  # Check if it's a CSV file
            print(f"Extracting: {file_name}")
            z.extract(file_name, output_dir)
            print(f"Saved to: {output_dir}/{file_name}")
```
